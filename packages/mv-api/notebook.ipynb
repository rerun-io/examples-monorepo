```python
import rerun as rr
from rerun.experimental.dataloader import DataSource, Field, FixedRateSampling, RerunIterableDataset, VideoFrameDecoder

client = rr.catalog.CatalogClient(url="127.0.0.1:9991")
```

```python
source = DataSource(
    dataset=client.get_dataset("assembly101"),
    segments=[
        "assembly101__all__nusar-2021_action_both_9026-c06e_9026_user_id_2021-02-03_170116"
    ],
)
```

```python
fields = {
    "exo_video": Field("/world/exo/C10095/pinhole/video:VideoStream:sample", decode=VideoFrameDecoder(codec="av1")),
}

ds = RerunIterableDataset(
    source=source,
    index="video_time",
    fields=fields,
    timeline_sampling=FixedRateSampling(rate_hz=10.0),
    shuffle=False,
)
```
